```python
#histogram

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)
scores = np.random.normal(70,10,100)
bins = [0,30,70,90,100]
plt.hist(scores,bins=bins,color='skyblue',edgecolor='darkgreen')
plt.xlabel("scores")
plt.ylabel("# of students")
plt.title("Distribution of marks for students",color='blue')
plt.show()
```

```python
#multiple data set

legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

#Transaction Amount Distribution - Legit vs fraud
plt.hist(legit_transactions,color='green',bins=10,label='fraud',edgecolor='black',alpha=0.5)
plt.hist(fraud_transactions,color='red',bins=10,label='legit',edgecolor='black',alpha=0.5)


plt.title("Transaction amount distribution - legit vs fraud",color='purple')
plt.xlabel("Amount (in S)")
plt.ylabel('Frequancy')
plt.legend(loc='upper right')
plt.grid(axis='y')
```

```python
# axv line
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

plt.hist(marks,color='skyblue',edgecolor='black',bins=bins)

plt.axvline(x=33,
            label='passing marks (33)',
            color='r',
            linewidth='2',
            linestyle='dotted')
plt.ylabel("number of students")
plt.xlabel("Marks")
plt.title('distribution of marks for students')
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.random.normal(50, 10, 10000)
y = x * 0.5 + np.random.normal(0, 5, 10000)

plt.hist2d(x,y,bins=30,cmap='Blues')
plt.colorbar()
plt.show()
```

```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt

iris = load_iris()

x = iris.data[:,0] # sepal length
y = iris.data[:,1] # sepal width

plt.hexbin(x,y,gridsize=15,cmap='Blues')
```

```python
height = np.random.randint(150,195,100)
weight = np.random.randint(55,98,100)

plt.hist2d(height,weight,bins=10,cmap='Blues')
plt.colorbar()
plt.show()
```

```python
import seaborn as sns
import pandas as pd

titanic = sns.load_dataset('titanic')

# titanic.head(50)

titanic['age'] = titanic['age'].fillna(
    titanic['age'].median()

)

titanic['deck'] = titanic['deck'].fillna(
    titanic['deck'].mode()[0]
)

titanic.isnull().sum()
```

```python
plt.hist2d(titanic['age'],titanic['fare'],bins=10,cmap='Blues')
plt.colorbar()
plt.xlabel("age")
plt.ylabel("fare")
plt.title("Age vs Fare")

plt.show()
```

```python
plt.hist2d(titanic['age'],titanic['fare'],bins=20,cmap='Blues')
plt.show()
```

```python
plt.hexbin(
    titanic["age"],
     titanic["fare"],
    gridsize=20,
    cmap="plasma"
)

plt.colorbar()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 11)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(x, x)
plt.title("Line Plot")

plt.subplot(1,2,2)
plt.bar(x, x)
plt.title("Bar Plot")

plt.show()
```
